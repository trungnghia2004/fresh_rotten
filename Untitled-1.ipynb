```python
import numpy as np
import json
from tensorflow import keras
from tensorflow.keras.preprocessing import image

# ===== CONFIG =====
MODEL_PATH = r"F:\fresh_rotten\cnn_best.keras"  
IMG_SIZE = (224, 224)

# ===== LOAD MODEL =====
model = keras.models.load_model(MODEL_PATH)
print("Model loaded successfully!")

# ===== LOAD CLASS NAMES =====
# CÃƒÂ¡ch 1: NÃ¡ÂºÂ¿u cÃƒÂ³ file class_names.json
try:
    with open(r"F:\fresh_rotten\class_names.json", "r") as f:
        class_names = json.load(f)
    print("Loaded class_names from file.")
except:
    print("No class_names.json found.")
    class_names = None

# ===== PREPROCESS FUNCTION =====
def preprocess(img_path):
    img = image.load_img(img_path, target_size=IMG_SIZE)
    arr = image.img_to_array(img).astype("float32") / 255.0
    arr = np.expand_dims(arr, 0)
    return arr

# ===== DECODE FUNCTION =====
def decode_label(label):
    if "_" not in label:
        return "unknown", "unknown"
    fruit, quality = label.rsplit("_", 1)
    return fruit, quality

# ===== PREDICT FUNCTION =====
def predict_image(img_path):
    x = preprocess(img_path)
    probs = model.predict(x)[0]

    idx = int(np.argmax(probs))
    confidence = float(np.max(probs))

    if class_names is not None:
        label = class_names[idx]
        fruit, quality = decode_label(label)
    else:
        fruit, quality = "unknown", "unknown"

    print("\n===== RESULT =====")
    print("Fruit:", fruit)
    print("Quality:", quality)
    print("Confidence:", round(confidence, 4))

# ===== RUN TEST =====
if __name__ == "__main__":
    img_path = r"F:\fresh_rotten\test\banana_fresh_000003.png"   # Ã„â€˜Ã¡Â»â€¢i tÃƒÂªn Ã¡ÂºÂ£nh
    predict_image(img_path)
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from tensorflow import keras
from tensorflow.keras.preprocessing import image

TEST_ROOT = r"C:\Users\ADMIN\Downloads\processed_split_2level\test"
IMG_SIZE = (224, 224)

# Because your model output is (None, 1)
class_names = ["fresh", "rotten"]


def collect_test_samples(test_root):
    items = []

    if not os.path.isdir(test_root):
        raise FileNotFoundError(f"Test folder not found: {test_root}")

    for fruit in sorted(os.listdir(test_root)):
        fruit_dir = os.path.join(test_root, fruit)

        if not os.path.isdir(fruit_dir):
            continue

        for quality in ["fresh", "rotten"]:
            q_dir = os.path.join(fruit_dir, quality)

            if not os.path.isdir(q_dir):
                continue

            for fn in os.listdir(q_dir):
                p = os.path.join(q_dir, fn)

                if os.path.isfile(p):
                    items.append((p, quality))  # only fresh/rotten

    return items


def preprocess(img_path):
    img = image.load_img(img_path, target_size=IMG_SIZE)
    arr = image.img_to_array(img).astype("float32") / 255.0
    return np.expand_dims(arr, 0)


def evaluate_model(model_path, class_names, test_items):

    model = keras.models.load_model(model_path)

    model_num_classes = int(model.output_shape[-1])
    expected_num_classes = len(class_names)

    # Binary case (your case)
    if model_num_classes == 1 and expected_num_classes == 2:
        is_binary = True
        print("Detected binary model (1 output neuron).")
    elif model_num_classes == expected_num_classes:
        is_binary = False
        print("Detected multi-class model.")
    else:
        raise ValueError(
            f"Class count mismatch for {os.path.basename(model_path)}: "
            f"model outputs {model_num_classes}, but class_names has {expected_num_classes}."
        )

    label_to_idx = {c: i for i, c in enumerate(class_names)}
    labels = list(range(expected_num_classes))

    y_true = []
    y_pred = []

    for img_path, true_label in test_items:

        x = preprocess(img_path)
        probs = model.predict(x, verbose=0)[0]

        if is_binary:
            pred_idx = int(probs > 0.5)
        else:
            pred_idx = int(np.argmax(probs))

        y_pred.append(pred_idx)
        y_true.append(label_to_idx[true_label])

    if len(y_true) == 0:
        raise ValueError("No test samples found.")

    acc = accuracy_score(y_true, y_pred)
    cm = confusion_matrix(y_true, y_pred, labels=labels)

    report = classification_report(
        y_true,
        y_pred,
        labels=labels,
        target_names=class_names,
        digits=4,
        zero_division=0
    )

    print(f"\n===== {os.path.basename(model_path)} =====")
    print(f"Accuracy: {acc:.4f}")
    print("Classification report:")
    print(report)
    print("Confusion matrix:")
    print(cm)

    plt.figure(figsize=(6, 5))
    plt.imshow(cm, interpolation="nearest", cmap="Blues")
    plt.title(f"Confusion Matrix - {os.path.basename(model_path)}")
    plt.colorbar()

    ticks = np.arange(len(class_names))
    plt.xticks(ticks, class_names)
    plt.yticks(ticks, class_names)

    plt.ylabel("True label")
    plt.xlabel("Predicted label")

    plt.tight_layout()
    plt.show()


# ================= RUN =================

test_items = collect_test_samples(TEST_ROOT)
print(f"Total test samples: {len(test_items)}")
print("Unique labels:", sorted(set(label for _, label in test_items)))

model_paths = [
    r"F:\fresh_rotten\cnn_best.keras",
    r"F:\fresh_rotten\mobilenet_fruit_quality.keras",
]

for mp in model_paths:
    evaluate_model(mp, class_names, test_items)
```
